```python
def make_averager():
    series = []
    
    def averager(val: float):
        series.append(val)
        return sum(series)/len(series)
    
    return averager

avg = make_averager()

print(avg(10))
print(avg.__annotations__)
print(avg.__code__.co_freevars)
print(avg.__closure__)
print(avg.__closure__[0].cell_contents)

print(avg(11))
print(avg(12))
print(avg.__closure__[0].cell_contents)
```
